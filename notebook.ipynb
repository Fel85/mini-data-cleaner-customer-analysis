```python
import pandas as pd

df = pd.read_csv("data/customers.csv", sep="\t") # Cargar el archivo CSV con tabulaciones como separadores
print(df.head()) # Mostrar las primeras filas del DataFrame
df.head()
```

```python
df.columns = df.columns.str.strip() # Eliminar espacios en blanco de los nombres de las columnas
```

```python
df["Education"] = df["Education"].str.strip().str.title() # Limpiar la columna "Education" eliminando espacios y capitalizando las palabras
df["Marital_Status"] = df["Marital_Status"].str.strip().str.title() # Limpiar la columna "Marital_Status" eliminando espacios y capitalizando las palabras
```

```python
df["Dt_Customer"] = pd.to_datetime(df["Dt_Customer"], format='%d-%m-%Y') # Convertir la columna "Dt_Customer" a formato de fecha, manejando errores
```

```python
df.isnull().sum() # Verificar nuevamente la cantidad de valores nulos después de la limpieza
```

```python
df["Income"] = df["Income"].fillna(df["Income"].median()) # Rellenar los valores nulos en la columna "Income" con la mediana de esa columna
```

```python
df = df.drop_duplicates() # Eliminar filas duplicadas del DataFrame
```

```python
df["Age"] = 2024 - df["Year_Birth"] # Crear una nueva columna "Age" calculando la edad a partir del año de nacimiento
```

```python
df["Customer_Since"] = (pd.Timestamp("today") - df["Dt_Customer"]).dt.days # Crear una nueva columna "Customer_Since" calculando el número de días desde que el cliente se unió 
```

```python
print("Filas originales:", len(df))
print("Duplicados:", df.duplicated().sum())
print("Nulos:\n", df.isnull().sum())
```

```python
df.to_csv("data/clean.csv", index=False) # Guardar el DataFrame limpio en un nuevo archivo CSV sin incluir el índice
```

```python
df["Age"] = 2024 - df["Year_Birth"] # Crear una nueva columna "Age" calculando la edad a partir del año de nacimiento
df["Age"].describe()
```

```python
df = df[df["Age"] < 100] # Filtrar el DataFrame para incluir solo clientes mayores de 18 años
```

```python
df["Income"].describe()
```

```python
Q1 = df["Income"].quantile(0.25) # Calcular el primer cuartil (Q1) de la columna "Income"
Q3 = df["Income"].quantile(0.75) # Calcular el tercer

IQR = Q3 - Q1 # Calcular el rango intercuartílico (IQR)

limite_inferior = Q1 - 1.5 * IQR # Calcular el límite inferior para detectar outliers
limite_superior = Q3 + 1.5 * IQR # Calcular el límite superior para detectar outliers

df = df[(df["Income"] >= limite_inferior) & (df["Income"] <= limite_superior)] # Filtrar el DataFrame para eliminar outliers en la columna "Income"
```

```python
df["Marital_Status"].value_counts() # Contar la cantidad de clientes por estado civil para verificar la limpieza de esa columna
```

```python
df["Marital_Status"] = df["Marital_Status"].replace({
    "Alone": "Single",
    "YOLO": "Single",
    "Absurd": "Single"
}) # Reemplazar valores inconsistentes en la columna "Marital_Status" para unificar las categorías
```

```python
df["Total_Spent"] = (
    df["MntWines"] +
    df["MntFruits"] +
    df["MntMeatProducts"] +
    df["MntFishProducts"] +
    df["MntSweetProducts"] +
    df["MntGoldProds"]
) # Crear una nueva columna "Total_Spent" sumando las cantidades gastadas en diferentes categorías de productos
```

```python
df["Has_Children"] = df["Kidhome"] + df["Teenhome"] # Crear una nueva columna "Has_Children" sumando las columnas "Kidhome" y "Teenhome" para indicar si el cliente tiene hijos
```

```python
df.groupby("Education")["Income"].mean()# Calcular el ingreso promedio por nivel educativo para analizar la relación entre educación e ingresos     
```
